# Get counts of item metadata

```python
import json
from collections import defaultdict
```

```python
def sum_json_fields(json_file_path):
    """
    Iterates over a JSON file containing a list of JSON objects
    and sums the values of common fields.

    Args:
        json_file_path (str): The path to the input JSON file.

    Returns:
        dict: A dictionary where keys are the field names and
              values are the sum of those fields across all objects.
              Returns None if the file is not found or is not a list.
    """
    
    field_aggs = dict()

    try:
        with open(json_file_path, 'r') as f:
            data = json.load(f)

        if not isinstance(data, list):
            print(f"Error: The JSON file '{json_file_path}' does not contain a list.")
            return None

        for item in data:
            if isinstance(item, dict):
                for key, value in item.items():
                    
                    current_set = field_aggs.setdefault(key, set())

                    if isinstance(value, list):
                        current_set.update(set(value))
                    elif isinstance(value, str):
                        current_set.add(value)
                    elif isinstance(value, (int, float)):
                        pass
                    else:
                        pass
            else:
                print(f"Warning: Found an item in the list that is not a JSON object (dict): {item}")

        return field_aggs
        
    except FileNotFoundError:
        print(f"Error: The file '{json_file_path}' was not found.")
        return None
    except json.JSONDecodeError:
        print(f"Error: Could not decode JSON from the file '{json_file_path}'.")
        return None
    except Exception as e:
        print(f"An unexpected error occurred during aggregation: {e}")
        return None
```

```python
filename = "metadata.json"
field_aggs = sum_json_fields(filename)

for key, value in field_aggs.items():
    print(f"Key: {key}, Unique Count: {len(value)}")
```

# How each item's metadata is mapped

```python
import re

def print_item_data(json_file_path):
    """
    Iterates over a JSON file containing a list of JSON objects
    and print first 10 items

    Args:
        json_file_path (str): The path to the input JSON file.

    Returns:
        None
    """
    

    try:
        with open(json_file_path, 'r') as f:
            data = json.load(f)

        if not isinstance(data, list):
            print(f"Error: The JSON file '{json_file_path}' does not contain a list.")
            return None

        for item in data[:10]:
            item_id = item["video"]
            match = re.search(r'/(\d+)_', item_id)

            if match:
                item_id = match.group(1)
            else:
                item_id = ""
            print(f"Item ID: {item_id}")
            print(f"Video path: {item['video']}")
            print(f"Video prompt: {item['prompt']}")
            print(f"Image folder: images/{item_id}")
            print()

    except FileNotFoundError:
        print(f"Error: The file '{json_file_path}' was not found.")
        return None
    except json.JSONDecodeError:
        print(f"Error: Could not decode JSON from the file '{json_file_path}'.")
        return None
    except Exception as e:
        # Print the actual internal error for debugging
        print(f"An unexpected error occurred during aggregation: {e}")
        return None
```

```python
filename = "metadata_test.json"
print_item_data(filename)
```
